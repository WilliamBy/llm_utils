# Raw file datasets

```python
import json

# 读取并解析 dureader.jsonl 文件中的所有条目
dataset = []
with open('./triviaqa.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():  # 跳过空行
            item = json.loads(line)
            dataset.append(item)

print(f"length of dureader dataset: {len(dataset)}")
from random import choice
sample = choice(dataset)
sample
```

```python
prompt = f"Answer the question based on the given passage. Only give me the answer and do not output any other words. The following are some examples.\n\n{sample['context']}\n\n{sample['input']}",

from transformers import AutoTokenizer
model_path = "/home/yexuming/.cache/modelscope/hub/models/Lourdle/Llama-3-8B-Instruct-Gradient-1048k"

tokenizer = AutoTokenizer.from_pretrained(model_path)
inputs = tokenizer(prompt, return_tensors="pt")
prompt_tokens = inputs.input_ids.shape[-1]
print(f"prompt_tokens: {prompt_tokens}")
```

```python
from transformers import AutoModelForCausalLM
model = AutoModelForCausalLM.from_pretrained(model_path, dtype="float16").to("cuda:0")
inputs = inputs.to(model.device)
```

```python
outputs = model.generate(**inputs, max_new_tokens=1000)
```

```python
print(f"Generated tokens: {outputs.shape[-1]}")
gen_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
gen_text
```
