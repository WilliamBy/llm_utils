```python
import torch
from safetensors import safe_open
```

```python
def generate_random_tensors(num_tensors=3, shape=(2, 2)):
    """Generate a list of random tensors with specified shape."""
    return [torch.randn(*shape) for _ in range(num_tensors)]

tensors = {
    "single": torch.randn(2, 2),
    "multi": [torch.randn(2, 2) for _ in range(3)],
    "embed": {
        "child": torch.randn(2, 2)
    }
}

from safetensors.torch import save_file
save_file(tensors, "tensors.safetensors")
```

```python
tensors = {
    "a": torch.randn(2, 2),
    "b": torch.randn(2, 2),
}
from safetensors.torch import save_file
save_file(tensors, "tensors.safetensors")
```

```python
# load tensors
tensors = {}
with safe_open("tensors.safetensors", framework="pt", device=0) as f:
    print(f.keys())
    for k in f.keys():
        tensors[k] = f.get_tensor(k)
print(tensors)
```
