# Tensor Operations

## Basic Info

```python
import torch
t = torch.tensor([1, 2, 3])
t
t.shape
t.size()
t.numel()
t.ndim
```

## Relation between origin, cloned, detached and device

```python
if torch.cuda.is_available():
    t = t.to("cuda")
t.device
```

```python
print(t)
t.requires_grad
td = t.detach()
tc = td.cpu()
tn = tc.clone()
print(td, tc, tn)
```

```python
td[0] = 0
print(t, td, tc, tn)
```

```python
tc[1] = 0
print(t, td, tc, tn)
```

## Tensor Init

```python
from torch import Tensor, randn, zeros, ones, diag
print("Tensor: ", Tensor([1, 2, 3]))
print("randn: ", randn(2, 2))
print("zeros: ", zeros(2, 2))
print("ones: ", ones(2, 2))
print("diag: ", diag([1, 2, 3]))
```
