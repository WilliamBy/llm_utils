```python
import seaborn as sns
import numpy as np
from string import ascii_letters
import random
import pandas as pd

step2heat = {}
for i in ["a", "b", "c"]:
    arr = np.random.randint(1, 1000, size=100)
    step2heat[i] = arr

step2df = {}
for step, attn in step2heat.items():
    step2df[step] = pd.DataFrame({"step": step, "attention": attn, "pos": range(len(attn))})

df = pd.concat(list(step2df.values()))
sns.scatterplot(data=df, x="pos", y="attention", hue="step")
```
